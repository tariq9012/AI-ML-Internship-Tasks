# Task 1: Exploring and Visualizing a Simple Dataset

## Objective
Iris dataset ko load, inspect aur visualize karna taake data ke trends aur distributions samajh sakein.

## Dataset
Iris Dataset - seaborn library se load hoga (CSV download ki zaroorat nahi)

```python
# Libraries import karna
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Dataset load karna
df = sns.load_dataset('iris')

# Basic inspection
print("Shape:", df.shape)
print("\nColumn Names:", df.columns.tolist())
print("\nFirst 5 Rows:")
df.head()
```

```python
# Summary statistics
print("Info:")
print(df.info())
print("\nDescribe:")
df.describe()
```

```python
# Scatter Plot
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='sepal_length', y='petal_length', hue='species')
plt.title("Scatter Plot - Sepal vs Petal Length")
plt.show()
```

```python
# Histogram
plt.figure(figsize=(8, 5))
df['sepal_length'].hist(bins=20, color='steelblue', edgecolor='black')
plt.title("Histogram - Sepal Length Distribution")
plt.xlabel("Sepal Length")
plt.ylabel("Frequency")
plt.show()
```

```python
# Box Plot
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x='species', y='petal_length')
plt.title("Box Plot - Petal Length by Species")
plt.show()
```

## Results aur Insights

- Iris dataset mein 150 rows aur 5 columns hain
- 3 species hain: Setosa, Versicolor, Virginica
- Scatter plot se pata chala ke Setosa baaki dono se alag hai
- Histogram se sepal length ki distribution normal hai
- Box plot se pata chala ke Virginica ki petal length sabse zyada hai
